```python
import pandas as pd
import matplotlib.pyplot as plt

# Corrected the file name to match the available files ('air_quality.csv' with underscore)
df = pd.read_csv('air_quality.csv')

# Corrected 'city_center' to 'city' based on the available data values
city_center_data = df[df['Location'] == 'city']

# Corrected column names to 'Air_Quallity_category' and 'PM2.5_level' to match the CSV headers
avg_pollution = df.groupby('Location')[['Air_Quallity_category', 'PM2.5_level']].mean(numeric_only=True)

plt.figure(figsize=(8,4))
plt.plot(city_center_data['Date'], city_center_data['Air_Quallity_category'], marker='o', color='red')
plt.title('chart 1 : air_Quality_Category Trend at city center')
plt.xlabel('Date')
plt.ylabel('Air Quality Index(AQI)')
plt.grid(True)
plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 Levels over Dates)
plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2.5_level'], color='purple', s=50)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Bar Chart (Comparing average PM 2.5 by location)
plt.figure(figsize=(8, 4))
avg_pollution['PM2.5_level'].plot(kind='bar', color='orange')
plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)
plt.show()
```

```python
#chart 4
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('air_quality.csv')

plt.figure(figsize=(8,4))
plt.hist(df['PM2.5_level'], bins=5, color='green', edgecolor='black')
plt.title('Chart 4: PM2.5 Level Distribution')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
```

```python
#char 5
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_level'].mean().plot(kind='bar', color='blue')
plt.title('char 5 :Average PM2.5 Levels by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5_level')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_level'].mean().plot(kind='barh', color='blue')
plt.title('Chart: Average PM2.5 Levels by Location')
plt.xlabel('Average PM2.5 Level')
plt.ylabel('Location')
plt.show()
```
